```python
plt.figure('figsize=(8,6)')
```

```python
#1. scatter plot

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time',
    size='size',
    palette='deep'
)
plt.title("Scatterplot of Total Bill vs Tip")
plt.show()
```

```python
#2. line plot

sns.lineplot(data=tips, x= 'size', y='total_bill', hue='sex',marker='o')
plt.title("Lineplot of Total Bill vs Size")
plt.show()
```

```python
sns.lineplot(data=tips, x='size', y='total_bill', hue='sex', ci=None, marker='o')
plt.title("Lineplot of Total Bill vs Size")
plt.show()
```

```python
#3.  bar plot

sns.barplot(data=tips, x='day', y='total_bill', hue = 'sex',palette='muted')
plt.title("Barplot of Total Bill by Day")
plt.show()
```

```python
# 4. Boxplot 

sns.boxplot(data=tips, x='day', y='tip', hue='smoker')
plt.title("Boxplot of Tips by Day and Smoker Status")
plt.show()
```

```python
# 5. violin plot --> violin plot is used to show the distribution of numerical data. The height shows the values, and the width shows where the data is more concentrated. In this graph, it shows the distribution of total bills for different days and compares Lunch and Dinner


sns.violinplot(data=tips, x='day', y='total_bill', hue='time', split=True, palette='pastel')
plt.title("Violin Plot of Total Bill by Day and Time")  
plt.show()
```

```python

#6. count plot 

sns.countplot(data=tips, x='day', hue='smoker', palette='dark')
plt.title("Count Plot of Days by Smoker Status")
plt.show()
```

```python
# 7. Histogram of total bill with KDE

sns.histplot(data=tips, x='total_bill', bins=20, kde=True , color='blue')
plt.title("Histogram of Total Bill with KDE")
plt.show()
```

```python
#8. pairplot -->This is a pairplot created using Seaborn. It is used to explore relationships between multiple numerical variables

sns.pairplot(tips, hue='sex', vars=["total_bill", "tip", "size"] ,palette='husl')
plt.suptitle("pair plot: numerberic variables by gender" ,  y=1.02)
plt.show()
```

```python
# 10 catplot --> It helps us compare numerical values across categories.

sns.catplot(data=tips, x='day', y='tip', hue='sex', palette='bright', kind='point')
plt.title("catplot(point):Tips by day and gender")
plt.show()
```

```python

# 11. jointplot --> We choose a jointplot when we want to study the relationship between two numerical variables and also understand their individual distributions.

sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='smoker', color='purple', palette='coolwarm')
plt.suptitle("Jointplot: Total Bill vs Tip", y=1.02)
plt.show()
```
