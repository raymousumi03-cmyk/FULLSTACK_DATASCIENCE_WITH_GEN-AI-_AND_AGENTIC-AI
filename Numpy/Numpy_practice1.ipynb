```python
import numpy as np

# creating a Numpy array
num_array = np.arange(100).reshape(20, 5)
print(num_array)

# Display array in Fortran order 

print("\n  C type array:")
for num_array in np.nditer(num_array, order="C"):
	print(num_array,end=' ')
```

```python
import numpy as np

# creating a Numpy array
num_array = np.arange(100).reshape(20, 5)
print(num_array)

# Display array in Fortran order 

print("\n  F type array:")
for num_array in np.nditer(num_array, order="F"):
	print(num_array,end=' ')
```

## program for otp generation

```python
import random

def generate_otp(length=4):
    """Generate a numeric OTP of a specified length."""
    digits = '012345'
    otp = ''.join(random.choice(digits) for _ in range(length))
    return otp

# Example usage
otp_length = 4  # You can change this to any length you prefer
otp = generate_otp(otp_length)
print(f"Your OTP is: {otp}")
```

```python
# another code

import random

digits = "0123456789"

otp = ""

for i in range(6):
    otp = otp + random.choice(digits)

print("Your OTP is:", otp)
```
