```python
import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [2, 4, 5, 3, 6, 7, 4, 8],
    "Maths": [55, 70, 80, 65, 90, 95, 75, 98],
    "Science": [60, 75, 85, 70, 88, 92, 78, 96],
    "English": [65, 72, 82, 68, 85, 90, 80, 94]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
def generate_plot(plot_type):

  fig = plt.figure(figsize=(8,5))

  if plot_type == "Line Plot":
    plt.plot(df["Student"], df["Maths"], marker="o", label="Maths")
    plt.plot(df["Student"], df["Science"], marker="o", label="Science")
    plt.plot(df["Student"], df["English"], marker="o", label="English")

    plt.title("Student Marks Comparison")
    plt.xlabel("Student")
    plt.ylabel("Marks")
    plt.grid(True)
    plt.legend()

  elif plot_type == "Bar Chart": 

    x = np.arange(len(df['Student']))   

    plt.bar(x -0.25 , df['Maths'] , width=0.25 , label="Maths")

    plt.bar(x  , df['Science'] , width=0.25 , label="Science")

    plt.bar(x +0.25  , df['English'] , width=0.25 , label="English")

    plt.xticks(x , df['Student'])

    plt.title("All SUbject Comparison")
    plt.xlabel("Student")
    plt.ylabel("Marks")

    plt.legend()
    

  elif plot_type == "Study hr vs Maths marks":

   plt.scatter(df['Study_Hours'] , df['Maths'] ,s=100)

   plt.title("Study hr vs Maths marks")
   plt.xlabel("Stusy_hrs")
   plt.ylabel("Math marks")
  
   plt.grid(True)



  elif plot_type == "Study hr vs Science marks":
  
     plt.scatter(df['Study_Hours'] , df['Science'] ,s=100)
  
     plt.title("Study hr vs Science marks")
     plt.xlabel("Stusy_hrs")
     plt.ylabel("Science marks")
    
     plt.grid(True) 


 
  elif plot_type == "Study hr vs English marks":
  
     plt.scatter(df['Study_Hours'] , df['English'] ,s=100)
  
     plt.title("Study hr vs English marks")
     plt.xlabel("Stusy_hrs")
     plt.ylabel("English marks")
    
     plt.grid(True)     


  elif plot_type == "Maths Histogram":

   plt.hist(df['Maths'] , bins=5 , edgecolor='black')

   plt.title(" Maths mark Distribution")
   plt.xlabel("Math marks")
   plt.ylabel("no.of students")


  elif plot_type == "Subject box plot":

   plt.boxplot([df['English'], df['Maths'], df['Science'] ], tick_labels=['Maths' , 'Science' , 'English'])

   plt.title("Subject Marks Distribution")
   plt.ylabel("Marks")


  elif plot_type == "Subject total pie chart":

   maths_total = df["Maths"].sum()
   science_total = df["Science"].sum()
   english_total = df["English"].sum()  

   plt.pie([ maths_total , science_total , english_total], labels = ["Maths" , "Science", "English"], autopct="%1.1f%%")

   plt.title("Subject-wise Total Marks")


  plt.tight_layout()

  return fig 


# Gardio UI

demo = gr.Interface(
      fn = generate_plot ,
      inputs= gr.Radio(
             ["Bar Chart",
              "Study hr vs Maths marks",
              "Maths Histogram",
              "Subject box plot" , 
              "Subject total pie chart",
              "Study hr vs English marks",
              "Study hr vs Science marks"] ,
              
              label="choose Analysis" ),

             outputs=gr.Plot(label="Visualization"),

             title="Student Performance Analyzer",

             description="Select an analysis type to visualize student performance."
)

live= True

demo.launch()             





    
```
