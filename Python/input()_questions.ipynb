16. Write a program that checks if the input number is a prime number.

```python
n = int(input("Enter a number: "))

if n > 1:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
```

35. How do you validate if an entered input is a valid email address? 

```python
import re

email = input("Enter your email address: ")

pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'

if re.match(pattern, email):
    print("Valid email address")
else:
    print("Invalid email address")
```

43. Write a program to find the sum of all digits in a string entered by the user.

```python
s = input("Enter a string: ")

sum = 0

for i in s:
    if i.isdigit():
        sum += int(i)

print("Sum:", sum)
```

48. Write a program to count the number of consonants in a string entered by the 
user. 

```python
text = input("Enter a string: ")

count = 0

for i in text:
    if i.isalpha() and i not in "aeiouAEIOU":
        count += 1

print("Consonants:", count)
```

49. How do you check if a string entered by the user contains any punctuation?

```python
import string

text = input("Enter a string: ")

if any(i in string.punctuation for i in text):
    print("The string contains punctuation.")
else:
    print("The string does not contain punctuation.")
```

50. Write a program that accepts a sentence and prints the longest word.

```python
text = input("Enter a sentence: ")

words = text.split()

longest = max(words , key=len)

print("Longest word:", longest)
```
