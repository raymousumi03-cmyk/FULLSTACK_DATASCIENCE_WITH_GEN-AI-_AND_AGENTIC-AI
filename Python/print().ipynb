#f string method

```python
num1=90
num2=89
add=num1+num2

print(f'Addition of {num1} and {num2} is= {add}')
```

```python
name='Mousumi' 
age=23 
city='odisha' 
#hello my name is Mousumi and i am 20 year old from odisha

print(f'My name is {name} and i am {age} years old form {city}')
```

```python
num1=100 
num2=25 
num3=333 
avg=round((num1+num2+num3)/3,2)

print(f"The average of {num1} and {num2} and {num3} is= {avg} ")
```

```python
num1=10 
num2=20 
add = num1+ num2 
print('The addition of',num1,'and',num2,'is=',add) 
print('The addition of {} and {} is= {}'.format(num1,num2,add)) 
print(f'The addition of {num1} and {num2} is= {add}') 
```
