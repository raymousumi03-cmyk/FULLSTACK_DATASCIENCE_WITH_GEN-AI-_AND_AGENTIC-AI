## Tuple unpacking

```python
def add_sub_mul(x,y):
    c = x + y
    d = x - y
    e = x * y 
    return c,d,e 

add_sub_mul(5,6) # Python calculates and returns (11, -1, 30), but you don't store it anywhere.
                 # so we get o/p as a tuple  
```

```python
def greet():
    return 'hello'

def add_sub(x,y):
    c = x + y
    d = x - y
    return c,d # (11, -1)

r = greet()    
r1 = add_sub(5, 6) # stores the whole tuple (11, -1) in r1.

print(r)
print(r1) # (11, -1)

print(type(r))
print(type(r1)) # tuple
```

```python
def greet():
    return 'hello'

def add_sub(x,y):
    c = x + y
    d = x - y
    return c,d # (11, -1)

r = greet()    
r1 , r2 = add_sub(5, 6) # unpacks the tuple, putting each value into a separate variable.
                        # r1 = 11 & r2 = - 1 . so here the dtype is int. 
print(r)
print(r1) # 11
print(r2) # -1

print(type(r))
print(type(r1)) # int 
print(type(r2)) # int
```

```python
def add_sub_mul(x,y):
    c = x + y
    d = x - y
    e = x * y 
    return c,d,e 

r, r1,r2  = add_sub_mul(5,6) 

print(r)
print(r1)
print(r2)

print(type(r))
print(type(r1)) 
print(type(r2)) 
```

# return()

```python
def add(x, y):    
    c = x+y
    return c  # Take the value inside c and give it back to the place where the function was called.   
    
r = add(5, 6) # now here r = 11
print(r) # it will print 11
print(type(r)) # whose type will be int
```

# 1-> positional Argument

```python
def person(name, age):    
    print(name)
    print(age)    
person('Mousumi', 23)
```

```python
def person(name, age):    
    print(name)
    print(age)    
person('Mousumi') # here we need 2 parameter byt we have 1.
```

# Keyword Argument

```python
def person(name, age):    
    print(name)
    print(age+1)    
person(age = 23, name = 'Mousumi') # like this here we will pass keywords .
```

```python
def person(name, age, mob):  # we should take all as positional or keyword argument but we can't mix both.
    print(name)
    print(age+1)    
    print(mob)
person(age = 23, name = 'Mousumi', 2345)
```

```python
def person(name, age, mob, addr):    
    print(name)
    print(age+1)    
    print(mob)
    print(addr)
person(age = 23, name = 'Mousumi', mob=2345, addr = 'odisha')
```

# Default Argument

```python
def person(name, age):    
    print(name)
    print(age) 

person('vote') 
```

```python
def person(name, age=18):    # here we give a default value if we don't pass any value it won't show error it shows 18.
    print(name)
    print(age) 

person('vote') 
```
