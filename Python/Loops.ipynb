## While loop

```python
print("data science")
print("data science")
print("data science")
print("data science")
```

```python
i = 1

while i<=5:
    print("data science")
    i=i+1 #increment 
```

```python
i = 5

while i>=1:
    print("data science")
    i=i-1 #decrement
```

```python
i = 5

while i>=1:
    print("data science", i)
    i=i-1 #decrement
```

```python
i = 1

while i<=5:
    print("data science" , i)
    i=i+1 #increment 
```

## nested while

```python
i = 1

while i<=5:
    print("data science")

    j=1
    while j<=4:
        print("technology")
        j = j + 1

    i= i + 1   
    print() 
```

```python
i = 1

while i<=5:
    print("data science" , end=" - ")

    j=1
    while j<=4:
        print("technology" , end=" - ")
        j = j + 1

    i= i + 1   
    print() 
```

## break --> break the loop from the given condition

```python
for i in range (1,11):
    if i ==5:
        break
    print(i)   
```

```python
for i in range (1,11):
    if i ==10:
        break
    print(i)   
```

```python
for i in range (1,11):
    if i ==5:
        break
    print("Hello",i)   
```

## continue --> skip the specific condition and continue the loop


```python
for i in range (1,11):
    if i ==5:
        continue
    print(i)   
```

```python
for i in range (1,11):
    if i ==5:
        continue
    print("Hello",i)   
```

## practice questions with the help of above concepts

print numbers which are not divisible by 3 and 5.

```python
for i in range(1,5):
    if i<=5:
     print("# # # #")
```
