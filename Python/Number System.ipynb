#swap variable in python

```python
a=5
b=6

a=b
b=a

a,b=b,a

print(a)
print(b)# in above scenario we lost the value 5 
```

```python
a1=7 #we use one extra variable
b1=8

temp=a1
a1=b1
b1=temp

print(a1)
print(b1)
```

```python
#swap variable formulas

a=90
b=89
a=a+b
b=a-b
a=a-b
print(f"a {a}")
print(f"b {b}")
```

```python
a2=9
b2=5

a2,b2=b2,a2

print(f"value of a2={a2}")
print(f"value of b2={b2}")
```
