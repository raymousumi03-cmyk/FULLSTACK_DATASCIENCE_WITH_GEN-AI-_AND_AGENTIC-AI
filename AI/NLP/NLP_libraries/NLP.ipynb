## Stemming --> it do suffix stripping 


```python
from nltk.stem import PorterStemmer # doesn't work for all 
pst = PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
words_to_stem =['give','giving','given','gaved','thinking','loving','maximum','Hello']

for words in words_to_stem:
    print(words+ ':' + pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer  # sufffix stripping for all words
lst = LancasterStemmer()

for word in words_to_stem:
    print(word+ ':' + lst.stem(word))
    
```

```python
from nltk.stem import SnowballStemmer # mainly works as porter stemmer
sbst = SnowballStemmer('english')

for word in words_to_stem:
    print(word+ ':' + sbst.stem(word))
```

# Lemmatization -> converts a word into its original dictionary form, called a lemma.


```python
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
word_len = WordNetLemmatizer()
```

```python
words_to_stem
```

```python
for words in words_to_stem:
    print(words+ ':' + word_len.lemmatize(words))
```

#WordNetLemmatizer treats words as nouns. Therefore, it may not convert verb forms such as giving and thinking into their base forms.

# *To process them as verbs

```python
for word in words_to_stem:
    print(word + ': ' + word_len.lemmatize(word, pos='v'))
```

# Named Entity Recognition (NER)
# NER identifies names of people, organizations, locations, and other named entities in text

```python
from nltk import ne_chunk
```

```python
NE_sent = ' The US president stays in the WHITEHOUSE'
```

```python
NE_tokens = word_tokenize(NE_sent)
NE_tokens
```

```python
NE_tags = nltk.pos_tag(NE_tokens)
NE_tags
```

```python
NE_NER = ne_chunk(NE_tags)
print(NE_NER)
```
