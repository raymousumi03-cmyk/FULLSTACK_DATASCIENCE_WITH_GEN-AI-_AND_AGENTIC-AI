## python code to check odd and even

```python
x = 8
r = x % 2 

if r ==  0:
    print('even number')
    
```

```python
x = 9
r = x % 2 

if r ==  0:
    print('even number')
print('odd number')    
```

```python
x = 6
r = x % 2 

if r ==  0:
    print('even number')
if r != 0:
    print('odd number')  
```

## if else

```python
x = 9
r = x % 2 

if r ==  0:
    print('even number')
else:
    print('odd number') 
```

## Nested if-else

```python
x=3
r=x % 2

if r == 0:
    print("even number")
    if x>5:
        print("greater number")

else:
    print("odd number")
```

```python
x=4
r=x % 2

if r == 0:
    print("even number")
    
    if x>5:
        print("greater number")

else:
    print("odd number")
```

```python
x=4
r=x % 2

if r == 0:
    print("even number")
    
    if x>5:
        print("greater number")
    else:
        print("less number")

else:
    print("odd number")
```

## if-elif-else

```python
x = 3

if x == 1:
    print('one')
if x == 2:
    print('Two')
if x == 3:
    print('Three')
if x == 4:
    print('four')
```

```python
x = 3

if x == 1:
    print('one')
elif x == 2:
    print('Two')
elif x == 3:
    print('Three')
elif x == 4:
    print('four')
```

```python
x = 30

if x == 1:
    print('one')
elif x == 2:
    print('Two')
elif x == 3:
    print('Three')
elif x == 4:
    print('four')

else:
    print("number not found")
```
